```python
spec = load_environment("triple_t")
x = np.linspace(0, 2 * np.pi, 200)

fig, (left, right) = plt.subplots(1, 2, figsize=(9, 4))
for (x0, y0), (x1, y1) in spec.walls:
    left.plot([x0, x1], [y0, y1], color="black", lw=1)
left.set_aspect("equal")
left.set_xlabel("x (m)")
left.set_ylabel("y (m)")
left.set_title(f"{spec.name}  [{spec.hash}]")

right.plot(x, np.sin(x), label="sin")
right.plot(x, np.cos(x), label="cos")
right.set_xlabel("x")
right.set_ylabel("y")
right.legend()
right.set_title("sample")

fig.tight_layout()
# fig.savefig(FIGURES / "sample.png", dpi=300)
```
