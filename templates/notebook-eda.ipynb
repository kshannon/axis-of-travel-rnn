```python
%load_ext autoreload
%autoreload 2
%matplotlib inline
%config InlineBackend.figure_format = "retina"

import subprocess
from datetime import date
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scienceplots  # noqa: F401  registers the SciencePlots styles

import axis_of_travel_rnn
from axis_of_travel_rnn.blocks.geometry import geometry
from axis_of_travel_rnn.config import REPO_ROOT, section

plt.style.use(["science", "notebook", "no-latex"])
rng = np.random.default_rng(0)
FIGURES = Path("figures")

commit = subprocess.run(["git", "rev-parse", "--short", "HEAD"], capture_output=True,
                        text=True).stdout.strip() or "unknown"
print(f"axis-of-travel-rnn {axis_of_travel_rnn.__version__}  commit {commit}  {date.today()}")
print(f"environments: {', '.join(section('environments'))}")
```

```python
from axis_of_travel_rnn.config import DataConfig

LAB_RATE_HZ = 60  # every lab frame kept

eda = section("eda.default", REPO_ROOT / config_path)
plt.rcParams["figure.dpi"] = eda["figure_dpi"]

# Every block of the built folder in one split, as EDA reads the blocks whole and trains nothing.
data = DataConfig(blueprint=blueprint, rate_hz=LAB_RATE_HZ, split=(1.0, 0.0, 0.0), seed=0)
loaded = axis_of_travel_rnn.load_blocks(data)
blocks = loaded["blocks"]["train"]
print(f"{blueprint}, data id {loaded['data_id'][:12]}, {len(blocks)} blocks from {loaded['folder']}")
```

```python
track = geometry("triple_t")
x = np.linspace(0, 2 * np.pi, 200)

fig, (left, right) = plt.subplots(1, 2, figsize=(9, 4))
for (x0, y0), (x1, y1) in track["walls"]:
    left.plot([x0, x1], [y0, y1], color="black", lw=1)
left.set_aspect("equal")
left.set_xlabel("x (m)")
left.set_ylabel("y (m)")
left.set_title(track["name"])

right.plot(x, np.sin(x), label="sin")
right.plot(x, np.cos(x), label="cos")
right.set_xlabel("x")
right.set_ylabel("y")
right.legend()
right.set_title("sample")

fig.tight_layout()
# fig.savefig(FIGURES / "sample.png", dpi=300)
```
